# Pressure of a Quantum Gas

Transcribed from the handwritten lecture note **"Pressure of a quantum gas"**.

This notebook follows the derivation in the note and keeps the notation as close as possible to the original, while typesetting the equations cleanly.

## Setup

Consider a non-relativistic ideal quantum gas with Hamiltonian

$$
\hat H = \sum_i \frac{\hat p_i^2}{2m}
$$

in a box of volume $V = L^D$.

Classically, the equation of state is

$$
\beta P = \frac{N}{V}.
$$

We derive the quantum correction in the **grand canonical ensemble**.

## Single-particle spectrum

The single-particle eigenstates are labeled by wavevector $\mathbf{k}$, with

$$
\mathbf{k} = \frac{2\pi}{L}(i,j,k,\dots),
\qquad
\varepsilon_{\mathbf{k}} = \frac{\hbar^2 k^2}{2m}.
$$

For each mode $\mathbf{k}$, the grand partition function contributes

$$
\mathcal{Q}_{\mathbf{k}} =
\begin{cases}
1 + e^{-\beta(\varepsilon_{\mathbf{k}}-\mu)}, & \text{fermions},\\[4pt]
\left(1 - e^{-\beta(\varepsilon_{\mathbf{k}}-\mu)}\right)^{-1}, & \text{bosons}.
\end{cases}
$$

Therefore the full grand partition function is

$$
\mathcal{Q} = \prod_{\mathbf{k}} \mathcal{Q}_{\mathbf{k}}.
$$

## Grand potential

The grand potential is

$$
\Phi = -\frac{1}{\beta}\ln \mathcal{Q}.
$$

Equivalently,

$$
\Phi
= -\frac{1}{\beta}\sum_{\mathbf{k}}
\begin{cases}
\ln\!\left(1+e^{-\beta(\varepsilon_{\mathbf{k}}-\mu)}\right), & \text{fermions},\\[4pt]
-\ln\!\left(1-e^{-\beta(\varepsilon_{\mathbf{k}}-\mu)}\right), & \text{bosons}.
\end{cases}
$$

A compact way to write both cases is

$$
\Phi
= -\frac{1}{\beta}\sum_{\mathbf{k}}
\left(\mp \ln\!\left(1 \mp e^{-\beta(\varepsilon_{\mathbf{k}}-\mu)}\right)\right),
$$

with the upper sign for fermions and the lower sign for bosons.

## Dilute-gas expansion

Assume the gas is dilute, so that

$$
\beta(\varepsilon_{\mathbf{k}}-\mu) \gg 1.
$$

Then we may expand the logarithm using

$$
\ln(1+x)= -\sum_{m=1}^{\infty}\frac{(-x)^m}{m}.
$$

Applying this to the grand potential gives, mode by mode,

$$
\Phi_{\mathbf{k}}
=
\mp \frac{1}{\beta}\sum_{m=1}^{\infty}
\frac{(\pm)^m}{m}
e^{-\beta m(\varepsilon_{\mathbf{k}}-\mu)}.
$$

Summing over $\mathbf{k}$,

$$
\Phi
=
\mp \frac{1}{\beta}\sum_{\mathbf{k}}\sum_{m=1}^{\infty}
\frac{(\pm)^m}{m}
e^{-\beta m(\varepsilon_{\mathbf{k}}-\mu)}.
$$

## Continuum limit

For a large box,

$$
\sum_{\mathbf{k}}
\;\approx\;
\left(\frac{L}{2\pi}\right)^D \int d^Dk
=
V \int \frac{d^Dk}{(2\pi)^D}.
$$

So

$$
\Phi
=
\mp \frac{V}{\beta}
\int \frac{d^Dk}{(2\pi)^D}
\sum_{m=1}^{\infty}
\frac{(\pm)^m}{m}
e^{-\beta m(\varepsilon_{\mathbf{k}}-\mu)}.
$$

Using $\varepsilon_{\mathbf{k}}=\hbar^2k^2/(2m)$, the Gaussian integral over $k$ gives

$$
\Phi
=
-\frac{V}{\beta \lambda^D}
\sum_{m=1}^{\infty}
\frac{(\pm)^{m-1}}{m^{1+D/2}}e^{\beta m\mu},
$$

where the thermal wavelength is

$$
\lambda
=
\sqrt{\frac{2\pi\hbar^2\beta}{m}}.
$$

## Pressure and density

From the grand potential,

$$
P = -\frac{\Phi}{V},
\qquad
n \equiv \frac{N}{V}
= -\frac{1}{V}\frac{\partial \Phi}{\partial \mu}.
$$

Therefore,

$$
\beta P
=
\frac{1}{\lambda^D}
\sum_{m=1}^{\infty}
\frac{(\pm)^{m-1}}{m^{1+D/2}}e^{\beta m\mu},
$$

and

$$
n
=
\frac{1}{\lambda^D}
\sum_{m=1}^{\infty}
\frac{(\pm)^{m-1}}{m^{D/2}}e^{\beta m\mu}.
$$

These are the standard fugacity expansions, with fugacity

$$
z = e^{\beta\mu}.
$$

## Virial expansion

Recall the virial expansion

$$
\beta P
=
n\left(1 + B_2(T)n + B_3(T)n^2 + \cdots \right).
$$

Keeping only terms through order $z^2$:

$$
n\lambda^D = z \pm \frac{z^2}{2^{D/2}} + \cdots
$$

and

$$
\beta P\,\lambda^D
=
z \pm \frac{z^2}{2^{1+D/2}} + \cdots.
$$

Matching these gives the second virial coefficient

$$
B_2(T)
=
\mp \frac{\lambda^D}{2^{1+D/2}}.
$$

So:

- **bosons**: $B_2(T) < 0$,
- **fermions**: $B_2(T) > 0$.

## Interpretation

For bosons, the pressure is **reduced** relative to the classical ideal gas.

For fermions, the pressure is **increased**.

Classically, for pairwise interactions one has

$$
B_2
=
-\frac12\int d^Dq\,
\left(e^{-\beta V(q)} - 1\right).
$$

This lets one interpret the quantum-statistical correction as if bosons/fermions had an effective short-range interaction. In the note, the quoted effective potential is

$$
\beta V(r) = \mp e^{-2\pi r^2/\lambda^2},
$$

so bosons behave **as if** they attract and fermions **as if** they repel over a range of order $\lambda$.

Of course, this is only an effective description: the origin is **quantum statistics**, not a real interaction.

## Quantum degeneracy

Quantum effects become important when

$$
\lambda^D \frac{N}{V} \sim 1,
$$

that is, when the thermal wavelength becomes comparable to the interparticle spacing.

This is the onset of **quantum degeneracy**.
