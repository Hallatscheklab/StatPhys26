# The Degenerate Fermi Gas

We consider a gas of fermions in the high-density, low-temperature regime where quantum statistics matter strongly. The key condition is that the thermal de Broglie wavelength is large compared with the interparticle spacing, or equivalently

$$
n \lambda_T^D \gg 1,
$$

where $n = N/V$ and

$$
\lambda_T = \frac{h}{\sqrt{2\pi m k_B T}}.
$$

Typical examples are:

- conduction electrons in metals,
- electrons in white dwarfs,
- neutrons in neutron stars.

In all of these systems, Pauli exclusion forces particles to occupy many distinct one-particle states even at very low temperature. This leads to a finite pressure already at $T=0$, and to a low-temperature heat capacity that is linear in $T$.

Unless stated otherwise, the formulas below are written for a single internal species. If the particles carry a spin or flavor degeneracy $g_s$, multiply the momentum-space state counting and the density of states by $g_s$.


## 1. Occupation of states and the Fermi sea

For fermions, the occupation of a one-particle state with energy $\varepsilon$ is

$$
n_F(\varepsilon) = \frac{1}{e^{\beta(\varepsilon-\mu)}+1}.
$$

At $T=0$, this becomes a step function:

$$
n_F(\varepsilon) = \Theta(\mu-\varepsilon).
$$

At zero temperature the chemical potential equals the Fermi energy,

$$
\mu(T=0) = E_F,
$$

so that

$$
n_F(\varepsilon) = \Theta(E_F-\varepsilon).
$$

Thus all one-particle states with $\varepsilon < E_F$ are filled, and all states with $\varepsilon > E_F$ are empty.

For a translationally invariant gas with dispersion $\varepsilon_{\mathbf k}$, the filled region in $\mathbf{k}$-space is called the **Fermi sea**, and its boundary

$$
\varepsilon_{\mathbf k} = E_F
$$

is the **Fermi surface**.

For the nonrelativistic free gas,

$$
\varepsilon_{\mathbf k} = \frac{\hbar^2 k^2}{2m},
$$

so the Fermi surface is a sphere of radius $k_F$:

$$
E_F = \frac{\hbar^2 k_F^2}{2m}.
$$


## 2. Why the Fermi surface matters

At low but nonzero temperature, only states within an energy window of order $k_B T$ around the Fermi energy can change their occupation appreciably.

Excitations therefore consist of

- adding a particle above the Fermi surface, or
- removing a particle from below it, creating a **hole**.

The number of thermally active states is proportional to the volume of a thin shell around the Fermi surface:

$$
\#(\text{active states})
\sim
g(E_F)\,k_B T.
$$

This is the basic reason why the degenerate Fermi gas is gapless and why many thermodynamic quantities at low $T$ are controlled only by properties near the Fermi surface.


## 3. Zero-temperature pressure

For a classical ideal gas,

$$
P = \frac{N}{V} k_B T,
$$

so the pressure vanishes as $T \to 0$. For fermions this is not true.

Using thermodynamics,

$$
P = -\left(\frac{\partial F}{\partial V}\right)_{N,T}.
$$

At $T=0$, the entropy vanishes for this noninteracting ground state, so $F=E$ and hence

$$
P = -\left(\frac{\partial E}{\partial V}\right)_N.
$$

We now compute this for a free nonrelativistic Fermi gas in $D$ spatial dimensions.


## 4. Relation between density and Fermi momentum

At $T=0$, all states with $|\mathbf{k}| < k_F$ are occupied. For a single internal species,

$$
\frac{N}{V} = \int \frac{d^D k}{(2\pi)^D}\,\Theta(k_F-k).
$$

It is convenient to define

$$
\gamma_n := \int_{|\mathbf{k}|<1} \frac{d^D k}{(2\pi)^D} k^n,
$$

so that

$$
\frac{N}{V} = \gamma_0 k_F^D,
\qquad
k_F = \left(\frac{N/V}{\gamma_0}\right)^{1/D}.
$$

Equivalently, carrying out the angular integral explicitly,

$$
\frac{N}{V} = \frac{S_{D-1}}{D(2\pi)^D} k_F^D,
$$

where $S_{D-1}$ is the surface area of the unit sphere in $D$ dimensions. If there is an internal degeneracy $g_s$, this becomes

$$
n = \frac{N}{V} = g_s\,\frac{S_{D-1}}{D(2\pi)^D} k_F^D.
$$


## 5. Ground-state energy

The total energy at $T=0$ is

$$
E = \sum_{|\mathbf{k}|<k_F} \varepsilon_{\mathbf{k}}
= V \int_{|\mathbf{k}|<k_F} \frac{d^D k}{(2\pi)^D} \frac{\hbar^2 k^2}{2m},
$$

so that

$$
\frac{E}{V} = \frac{\hbar^2}{2m} \gamma_2 k_F^{D+2}.
$$

Since $k_F \propto (N/V)^{1/D}$, this implies

$$
\frac{E}{V} \propto \left(\frac{N}{V}\right)^{1+2/D}.
$$

For the free gas one can evaluate the ratio explicitly and obtain

$$
\frac{E}{N} = \frac{D}{D+2} E_F.
$$


## 6. Pressure at $T=0$

Because $E \propto V^{-2/D}$ at fixed $N$, differentiating with respect to $V$ gives

$$
P = -\left(\frac{\partial E}{\partial V}\right)_N = \frac{2}{D} \frac{E}{V}.
$$

Using the explicit integrals, one also finds

$$
P = \frac{2}{D+2}\,\frac{N}{V} E_F = \frac{2}{D+2} n E_F.
$$

This is the degeneracy pressure of the ideal Fermi gas.

In $D=3$,

$$
P = \frac{2}{5} n E_F.
$$

This remains finite even at $T=0$, unlike the classical gas. Physically, the pressure comes from Pauli exclusion: confining the fermions to a smaller volume forces occupation of larger momenta and therefore larger kinetic energies.


```{admonition} SI: Scaling Argument for the Chandrasekhar Mass
:class: note

A white dwarf is supported by **electron degeneracy pressure** against its self-gravity. A simple scaling argument shows why there is a maximum possible mass.

Let the star have mass $M$ and radius $R$. If the matter contains about one electron per baryon mass $\mu_e m_p$, then the number of electrons is

$$
N_e \sim \frac{M}{\mu_e m_p},
$$

so the electron number density scales as

$$
n_e \sim \frac{N_e}{R^3} \sim \frac{M}{\mu_e m_p R^3}.
$$

For **ultrarelativistic** electrons, the Fermi energy scales as $E_F \sim p_F c \sim \hbar c\, n_e^{1/3}$, so the degeneracy pressure scales like

$$
P_{\rm deg} \sim n_e E_F \sim \hbar c\, n_e^{4/3}.
$$

On the other hand, hydrostatic balance requires a pressure of order

$$
P_{\rm grav} \sim \frac{G M^2}{R^4},
$$

which you can estimate from gravitational binding energy $E_{\rm grav} \sim -GM^2/R$ divided by volume $R^3$.

Equating these two pressures gives

$$
\hbar c\left(\frac{M}{\mu_e m_p R^3}\right)^{4/3}
\sim
\frac{G M^2}{R^4}.
$$

Now the important point is that the factors of $R$ cancel:

$$
\frac{\hbar c}{(\mu_e m_p)^{4/3}} M^{4/3}
\sim
G M^2.
$$

So instead of determining a radius, the balance determines a **mass scale**,

$$
M_{\rm Ch} \sim \frac{(\hbar c/G)^{3/2}}{\mu_e^2 m_p^2}.
$$

Up to numerical factors of order one, this is the **Chandrasekhar mass**. For a carbon-oxygen white dwarf, $\mu_e \approx 2$, giving a mass of order $1.4\,M_\odot$.

This cancellation is special to the relativistic equation of state $P \propto n^{4/3}$. In the nonrelativistic regime one has instead $P \propto n^{5/3}$, and then the pressure balance still depends on $R$, so one gets a mass-radius relation rather than a limiting mass.
```


## 7. Density of states formulation

To study finite temperature, it is useful to introduce the density of one-particle states per unit volume,

$$
g(E) = \int \frac{d^D k}{(2\pi)^D}\,\delta(E-\varepsilon_{\mathbf{k}}).
$$

If there is an internal degeneracy $g_s$, then the full density of states is $g_s g(E)$.

Then momentum integrals can be rewritten as energy integrals:

$$
\int \frac{d^D k}{(2\pi)^D}\,f(\varepsilon_{\mathbf{k}}) = \int dE\,g(E)\,f(E).
$$

In particular,

$$
\frac{N}{V} = \int dE\,g(E)\,n_F(E),
$$

and

$$
\frac{E}{V} = \int dE\,g(E)\,E\,n_F(E).
$$

Here $n_F(E) = 1/(e^{\beta(E-\mu)}+1)$, and one may take the integrals over all energies with the understanding that $g(E)=0$ outside the physical band. Low-temperature thermodynamics is then controlled by the behavior of $g(E)$ near $E=\mu \approx E_F$.


## 8. Sommerfeld expansion

At $T=0$, the Fermi function is a step function. At low but finite $T$, it is a smoothed step, and one can expand integrals of the form

$$
\int dE\, f(E)\, n_F(E)
$$

systematically in powers of temperature.

Before writing the formula, it is useful to see why the first correction has the form it does. The difference between the finite-$T$ Fermi function and the zero-temperature step function is localized in an energy window of width

$$
\Delta E \sim k_B T
$$

around $E=\mu$. So the correction to the integral comes only from this narrow region near the Fermi surface.

Now expand $f(E)$ about $E=\mu$:

$$
f(E) = f(\mu) + (E-\mu) f'(\mu) + \cdots.
$$

The term proportional to $f(\mu)$ does not contribute at leading order, because the smoothed Fermi function removes about as much weight just below $\mu$ as it adds just above $\mu$; the net area difference relative to the sharp step is zero. Therefore the first nonvanishing contribution comes from the fact that $f(E)$ varies across the thermal window, namely from the slope term $(E-\mu)f'(\mu)$.

Since this slope term is of size

$$
(E-\mu) f'(\mu) \sim (k_B T) f'(\mu),
$$

and it contributes over a window of width $\Delta E \sim k_B T$, the total correction scales as

$$
\delta \! \int dE\, f(E) n_F(E)
\sim
(k_B T)^2 f'(\mu).
$$

This explains both why the leading correction is quadratic in $T$ and why it is controlled by the derivative $f'(\mu)$ rather than by $f(\mu)$ itself.

The Sommerfeld expansion states that

$$
\int dE\, f(E)\, n_F(E)
= \int_{-\infty}^{\mu} dE\, f(E)
+ \frac{\pi^2}{6}(k_B T)^2 f'(\mu)
+ O(T^4).
$$

A remarkable feature is that the leading correction depends only on $f'(\mu)$, not on the detailed form of $f(E)$ far from the Fermi surface.

Applying this to the particle density gives

$$
\frac{N}{V} = \int_{-\infty}^{\mu} dE\, g(E)
+ \frac{\pi^2}{6}(k_B T)^2 g'(\mu)
+ O(T^4),
$$

and applying it to the energy density gives

$$
\frac{E}{V} = \int_{-\infty}^{\mu} dE\, E g(E)
+ \frac{\pi^2}{6}(k_B T)^2 \left.\frac{d}{dE}\big(E g(E)\big)\right|_{E=\mu}
+ O(T^4).
$$

At fixed particle density one must let $\mu$ acquire a weak temperature dependence. To leading order,

$$
\mu(T) = E_F - \frac{\pi^2}{6}(k_B T)^2 \frac{g'(E_F)}{g(E_F)} + O(T^4).
$$


## 9. Low-temperature heat capacity

We want the heat capacity at fixed particle number,

$$
C_{V,N} = \left(\frac{\partial E}{\partial T}\right)_{V,N}.
$$

Because both $E$ and $N$ depend on $\mu$ and $T$, one must account for the weak temperature dependence of $\mu(T)$ needed to keep $N$ fixed.

Start from the Sommerfeld expansion for the energy density,

$$
\frac{E}{V} = \int_{-\infty}^{\mu} dE\, E g(E)
+ \frac{\pi^2}{6}(k_B T)^2 \left.\frac{d}{dE}\big(E g(E)\big)\right|_{E=\mu}
+ O(T^4).
$$

Now expand this to leading order around $\mu = E_F$. The first term changes because the upper limit shifts from $E_F$ to $\mu(T)$:

$$
\int_{-\infty}^{\mu} dE\, E g(E)
=
\int_{-\infty}^{E_F} dE\, E g(E)
+ E_F g(E_F)\,[\mu(T)-E_F]
+ O(T^4),
$$

since $\mu(T)-E_F = O(T^2)$. In the explicit $T^2$ term we may set $\mu = E_F$ to this order, so

$$
\frac{E}{V}
=
\frac{E_0}{V}
+ E_F g(E_F)\,[\mu(T)-E_F]
+ \frac{\pi^2}{6}(k_B T)^2 \left.\frac{d}{dE}\big(E g(E)\big)\right|_{E=E_F}
+ O(T^4).
$$

Using the result from the previous section,

$$
\mu(T)-E_F = - \frac{\pi^2}{6}(k_B T)^2 \frac{g'(E_F)}{g(E_F)} + O(T^4),
$$

this becomes

$$
\frac{E}{V}
=
\frac{E_0}{V}
- \frac{\pi^2}{6}(k_B T)^2 E_F g'(E_F)
+ \frac{\pi^2}{6}(k_B T)^2 \big[g(E_F)+E_F g'(E_F)\big]
+ O(T^4).
$$

The terms proportional to $E_F g'(E_F)$ cancel, leaving

$$
\frac{E}{V} = \frac{E_0}{V} + \frac{\pi^2}{6}(k_B T)^2 g(E_F) + O(T^4).
$$

Differentiating at fixed $N$ and $V$ then gives

$$
C_{V,N}
=
V\left(\frac{\partial}{\partial T}\frac{E}{V}\right)_{N}
=
V\,\frac{\pi^2}{3}\,k_B^2 T\, g(E_F)
+ O(T^3).
$$

So the heat capacity is **linear in temperature**:

$$
C_V \propto T.
$$

This is one of the most important results for metals and more generally for degenerate fermionic matter. If an internal degeneracy $g_s$ is present, replace $g(E_F)$ by the full density of states $g_s g(E_F)$.


## 10. Physical interpretation

The linear law

$$
C_V = V\,\frac{\pi^2}{3}\,k_B^2 T\, g(E_F)
$$

reflects the fact that only a fraction of order

$$
\frac{\#(\text{thermally active degrees of freedom})}{\#(\text{all particles})} \sim \frac{k_B T}{E_F}
$$

can be excited thermally. Most fermions are buried deep inside the Fermi sea and cannot move because nearby states are already occupied.

This should be contrasted with:

- **gapped systems**, where $C_V \sim e^{-\Delta/(k_B T)}$,
- **phonons in 3D**, where $C_V \sim T^3$.

Hence a degenerate Fermi gas has a much larger low-temperature heat capacity than a gapped system, while still vanishing as $T \to 0$.


## 11. Summary

For an ideal degenerate Fermi gas:

1. At $T=0$, all states below $E_F$ are filled.
2. The filled region in momentum space is the Fermi sea, bounded by the Fermi surface.
3. Even at $T=0$, the gas has a nonzero pressure,

$$
P = \frac{2}{D}\frac{E}{V} = \frac{2}{D+2} n E_F.
$$

4. At low temperature, thermodynamics is controlled only by states near the Fermi surface.
5. The specific heat is linear in $T$,

$$
C_V = V\,\frac{\pi^2}{3} k_B^2 T\, g(E_F).
$$

These are the central signatures of Fermi degeneracy.
