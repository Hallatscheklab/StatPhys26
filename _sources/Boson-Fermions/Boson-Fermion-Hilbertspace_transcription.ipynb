# Bosons, Fermions, and the Many-Body Hilbert Space




## Hilbert Space of Particles

Consider a single particle with Hilbert space spanned by position kets $\lvert x\rangle$.

Usually one takes
$x \in \mathbb{R}$,
but in AMO or condensed-matter settings one may also consider
$x \in \mathbb{Z}$, for example for atoms in an optical lattice.

This distinction will not matter much in what follows, except that one writes either integrals or sums:

$$
\mathbb{I} = \int dx\, \lvert x\rangle \langle x\rvert
\qquad \text{vs.} \qquad
\mathbb{I} = \sum_x \lvert x\rangle \langle x\rvert
$$

and likewise

$$
\langle x \mid y \rangle = \delta(x-y)
\qquad \text{vs.} \qquad
\langle x \mid y \rangle = \delta_{x,y}.
$$

In these notes we use the discrete notation.

For a state $\lvert \psi \rangle$, the wavefunction in this basis is

$$
\psi(x) = \langle x \mid \psi \rangle.
$$

For two distinguishable particles, the Hilbert space is spanned by

$$
\lvert x_1, x_2\rangle = \lvert x_1\rangle \otimes \lvert x_2\rangle,
$$

with wavefunction

$$
\psi(x_1,x_2)=\langle x_1,x_2 \mid \psi\rangle.
$$

Similarly, for $N$ distinguishable particles one has the wavefunction

$$
\psi(x_1,x_2,\dots,x_N)=\langle x_1,x_2,\dots,x_N \mid \psi\rangle.
$$

For distinguishable particles this wavefunction lives in the full tensor-product Hilbert space. For identical particles, however, the physical Hilbert space is the symmetric subspace (bosons) or antisymmetric subspace (fermions) of that tensor-product space.



## Exchange Symmetry for Identical Bosons

A basic physical fact is that if the particles belong to the same species, then exchanging their labels does not produce a new physical state. In first quantization, this means that physical states must lie in the symmetric subspace.

Equivalently,

$$
\hat{P}_{ij}\lvert \psi \rangle = \lvert \psi \rangle
$$

for every exchange operator $\hat{P}_{ij}$. Equivalently, in the position representation the bosonic wavefunction satisfies

$$
\psi(\dots,x_i,\dots,x_j,\dots)=\psi(\dots,x_j,\dots,x_i,\dots).
$$

This is **exchange symmetry**, and it has several important consequences.

### Consequences for observables

Any physical observable should be represented by an operator $\hat{O}$ that does not distinguish particle labels. Thus, for example,

$$
\langle x_1,x_2 \mid \hat{O} \mid x_1,x_2\rangle
=
\langle x_2,x_1 \mid \hat{O} \mid x_2,x_1\rangle.
$$

There is no way to measure *which* boson is at $x$; only that a boson is at $x$.

Equivalently, physical observables must respect exchange symmetry, i.e. they must commute with all exchange operators $\hat{P}_{ij}$.

For example, the local number operator $\hat{n}(x)$ counts how many particles occupy the single-particle state $x$. In a two-particle state its expectation value is schematically

$$
\langle \hat{n}(x) \rangle
=
\sum_{x_2}\psi^*(x,x_2)\psi(x,x_2)
+
\sum_{x_1}\psi^*(x_1,x)\psi(x_1,x).
$$



### Exchange and Interference

Exchange symmetry allows **constructive interference**.

In quantum mechanics, the amplitude to evolve from an initial configuration
$1,2$
to a final configuration
$1',2'$
is a sum over all paths connecting the initial and final states.

For bosons, amplitudes associated with exchanged particle labels contribute to the same physical final state, and therefore add with the same sign.

This has measurable consequences, for example in two-particle interference effects such as the **Hanbury-Brown--Twiss** effect.



## Bosons in First-Quantized Language

### Wavefunctions and Exchange Symmetry

In first quantization we consider wavefunctions

$$
\psi(x_1,x_2,\dots)
$$

and impose bosonic symmetry:

$$
\psi(\dots,x_i,\dots,x_j,\dots)
=
\psi(\dots,x_j,\dots,x_i,\dots).
$$

Equivalently, if $\hat{P}_{ij}$ denotes the operator that swaps particle labels $i \leftrightarrow j$, then

$$
\hat{P}_{ij}\psi = \psi.
$$

Is this consistent with time evolution? Yes, because any physically admissible Hamiltonian must preserve exchange symmetry:

$$
\hat{H} = \sum_i \frac{\hat{p}_i^2}{2m} + \sum_i V(\hat{x}_i) + \frac12 \sum_{i\neq j} U(\hat{x}_i-\hat{x}_j) + \cdots
$$

and

$$
\hat{P}_{ij} \hat{H} \hat{P}_{ij} = \hat{H}.
$$

Therefore, if

$$
i\partial_t \psi = \hat{H}\psi
$$

and $\hat{P}_{ij}\psi(0)=\psi(0)$, then $\hat{P}_{ij}\psi(t)=\psi(t)$ for all $t$.

For example, one must **forbid** terms like

$$
V_1(x_1)+V_2(x_2)
\qquad \text{with } V_1\neq V_2,
$$

because those distinguish particle 1 from particle 2.



## Bosonic Occupation-Number Basis

Because of exchange symmetry, it is natural to expand states using symmetrized basis states.

For two particles,

$$
\lvert x_1,x_2\rangle_+
=
\begin{cases}
\frac{1}{\sqrt{2}}
\left(\lvert x_1,x_2\rangle + \lvert x_2,x_1\rangle\right), & x_1\neq x_2,\\[1em]
\lvert x_1,x_1\rangle, & x_1=x_2.
\end{cases}
$$

More generally,

$$
\lvert \{x_i\}\rangle_+
=
\frac{1}{\sqrt{\mathcal N}}
\sum_{\text{perm}} \lvert \text{Perm}\{x_i\}\rangle,
$$

where the sum runs over all permutations of the list $\{x_i\}$ and $\mathcal N$ is the appropriate normalization factor.

Since the ordering no longer matters, we instead label states by the **occupation numbers** $n_x$, where $n_x$ is the number of particles occupying the single-particle state $x$.

For example, for $N=2$ particles and orbitals $x=1,2,3$:

$$
\lvert 1,3\rangle_+
\;\leftrightarrow\;
\lvert n_1=1,n_2=0,n_3=1\rangle
=
\lvert 101\rangle,
$$

$$
\lvert 1,2\rangle_+
\;\leftrightarrow\;
\lvert n_1=1,n_2=1,n_3=0\rangle
=
\lvert 110\rangle,
$$

$$
\lvert 1,1\rangle_+
\;\leftrightarrow\;
\lvert n_1=2,n_2=0,n_3=0\rangle
=
\lvert 200\rangle.
$$

The state $\lvert \{n_x\}\rangle$ is called an **occupation-number basis state**.

The total particle number operator is

$$
\hat{N} = \sum_x \hat{n}_x.
$$



## Dependence on the Single-Particle Basis

The occupation basis depends on the choice of single-particle basis.

Instead of the basis $\lvert x\rangle$, we could use another basis $\lvert k\rangle$:

$$
\lvert x\rangle \to \lvert k\rangle,
\qquad
\lvert x_1,x_2,\dots\rangle \to \lvert k_1,k_2,\dots\rangle,
\qquad
\lvert \{n_x\}\rangle \to \lvert \{n_k\}\rangle.
$$

Using, for example,

$$
\langle x \mid k\rangle = \frac{e^{ikx}}{\sqrt{L}},
$$

one can work out overlaps such as

$$
\langle x_1,x_2,\dots \mid k_1,k_2,\dots\rangle.
$$

Computing overlaps between bosonic occupation states in different bases can become nontrivial; mathematically this is related to matrix permanents and, in complexity-theory language, to the **boson sampling problem**.



## Noninteracting Bosons

Suppose the Hamiltonian is noninteracting:

$$
\hat{H} = \sum_{i=1}^N \left(\frac{\hat{p}_i^2}{2m}+V(\hat{x}_i)\right).
$$

Let the single-particle eigenstates satisfy

$$
\left(\frac{\hat{p}^2}{2m}+V(\hat{x})\right)\lvert \alpha\rangle = E_\alpha \lvert \alpha\rangle.
$$

Then the many-body Hilbert space may be described either in first-quantized form,
$\lvert \alpha_1,\alpha_2,\dots,\alpha_N\rangle$
or in occupation-number form,
$\lvert \{n_\alpha\}\rangle$.

In the occupation basis,

$$
\hat{H} = \sum_\alpha E_\alpha \hat{n}_\alpha.
$$

Equivalently,

$$
\hat{H} = \sum_\alpha E_\alpha \hat{a}_\alpha^\dagger \hat{a}_\alpha,
$$

where

$$
\hat{n}_\alpha \lvert n_0,n_1,n_2,\dots\rangle
=
n_\alpha \lvert n_0,n_1,n_2,\dots\rangle.
$$

Thus the noninteracting bosonic problem is formally equivalent to a collection of decoupled harmonic oscillators,
with oscillator energy $E_\alpha$ and occupation number $n_\alpha$.



## Bosonic Creation and Annihilation Operators

Define bosonic ladder operators $\hat{a}_\alpha^\dagger$ and $\hat{a}_\alpha$ by

$$
\hat{n}_\alpha = \hat{a}_\alpha^\dagger \hat{a}_\alpha,
$$

with commutation relations

$$
[\hat{a}_\alpha,\hat{a}_\beta^\dagger]=\delta_{\alpha\beta},
$$

with all other commutators vanishing. Their action on occupation states is

$$
\hat{a}_\alpha^\dagger \lvert \dots,n_\alpha,\dots\rangle
=
\sqrt{n_\alpha+1}\,
\lvert \dots,n_\alpha+1,\dots\rangle,
$$

$$
\hat{a}_\alpha \lvert \dots,n_\alpha,\dots\rangle
=
\sqrt{n_\alpha}\,
\lvert \dots,n_\alpha-1,\dots\rangle.
$$

Thus, $\hat{a}_\alpha^\dagger$ **creates** a boson in state $\alpha$, while
$\hat{a}_\alpha$ **annihilates** a boson in state $\alpha$.

```{note}
A quick derivation of the standard relations is as follows. If we define the action of $\hat{a}_\alpha^\dagger$ and $\hat{a}_\alpha$ on occupation states by the formulas above, then

$$
\hat{a}_\alpha^\dagger \hat{a}_\alpha \lvert \dots,n_\alpha,\dots\rangle
=
n_\alpha \lvert \dots,n_\alpha,\dots\rangle,
$$

so indeed

$$
\hat{n}_\alpha = \hat{a}_\alpha^\dagger \hat{a}_\alpha.
$$

Likewise,

$$
\hat{a}_\alpha \hat{a}_\alpha^\dagger \lvert \dots,n_\alpha,\dots\rangle
=
(n_\alpha+1) \lvert \dots,n_\alpha,\dots\rangle,
$$

and therefore

$$
[\hat{a}_\alpha,\hat{a}_\alpha^\dagger]\lvert \dots,n_\alpha,\dots\rangle
=
\lvert \dots,n_\alpha,\dots\rangle.
$$

Since this holds on every occupation-number basis state, we get

$$
[\hat{a}_\alpha,\hat{a}_\beta^\dagger] = \delta_{\alpha\beta}.
$$
```

Under a single-particle unitary change of basis,

$$
\lvert \alpha\rangle = \sum_x U_{\alpha x}\lvert x\rangle,
$$

the creation operators transform as

$$
\hat{a}_\alpha^\dagger = \sum_x U_{\alpha x}\, \hat{a}_x^\dagger.
$$

A noninteracting Hamiltonian can be written as

$$
\hat{H} = \sum_{x,y} \hat{a}_x^\dagger H_{xy} \hat{a}_y
=
\sum_\alpha E_\alpha \hat{a}_\alpha^\dagger \hat{a}_\alpha.
$$

Here $H_{xy} = \langle x \mid \hat{H}_{\mathrm{1p}} \mid y \rangle$ are the matrix elements of the single-particle Hamiltonian.



## Grand-Canonical Statistical Mechanics of Noninteracting Bosons

For

$$
\hat{H} = \sum_\alpha E_\alpha \hat{n}_\alpha,
$$

the grand-canonical partition function is simple because all number operators commute:

$$
[\hat{n}_\alpha,\hat{n}_\beta]=0.
$$

Then

$$
e^{-\beta(\hat{H}-\mu \hat{N})}
=
e^{-\beta\sum_\alpha (E_\alpha-\mu)\hat{n}_\alpha},
$$

and

$$
Z
=
\sum_{\{n_\alpha\}}
e^{-\beta\sum_\alpha (E_\alpha-\mu)n_\alpha}
=
\prod_\alpha
\left(\sum_{n=0}^{\infty} e^{-\beta n(E_\alpha-\mu)}\right)
=
\prod_\alpha
\frac{1}{1-e^{-\beta(E_\alpha-\mu)}}.
$$

This expression converges only if the chemical potential lies below the lowest single-particle energy, i.e. $\mu < \min_\alpha E_\alpha$.

Therefore,

$$
\langle n_\alpha\rangle
=
\frac{1}{e^{\beta(E_\alpha-\mu)}-1},
$$

which is the **Bose-Einstein statistics**.



## Fermions: Antisymmetry and Pauli Exclusion

For fermions, physical states lie in the antisymmetric subspace, so

$$
\hat{P}_{ij}\psi = -\psi,
$$

so for two particles

$$
\psi(x_1,x_2) = -\psi(x_2,x_1).
$$

The properly antisymmetrized state is

$$
\lvert x_1,x_2,\dots\rangle_-
=
\frac{1}{\sqrt{\mathcal N}}
\sum_{\text{perm}}
(-1)^p
\lvert \text{Perm}\{x_i\}\rangle,
$$

where $p$ is the parity of the permutation.

Example:

$$
\lvert 1,2\rangle_-
=
\frac{1}{\sqrt{2}}
\left(\lvert 1,2\rangle - \lvert 2,1\rangle\right).
$$

If any two coordinates are equal, the antisymmetrized state vanishes:

$$
\lvert \{x_i\}\rangle_- = 0
\qquad \text{if any } x_i=x_j.
$$

This is the **Pauli exclusion principle**.

Hence one may restrict to ordered representatives such as

$$
x_1 < x_2 < x_3 < \cdots
$$

and use an occupation basis again, but now

$$
n_x \in \{0,1\}.
$$



## Exclusion as Destructive Interference

Pauli exclusion can also be understood as a **destructive interference** phenomenon.

In a path-integral picture, two paths that exchange identical fermions acquire opposite signs, so when the final positions coincide the amplitudes cancel.

That is why the amplitude for fermions to end up at the same place vanishes.



## Grand-Canonical Statistical Mechanics of Noninteracting Fermions

For noninteracting fermions,

$$
\hat{H}=\sum_\alpha E_\alpha \hat{n}_\alpha,
\qquad n_\alpha=0,1.
$$

Thus

$$
Z
=
\sum_{\{n_\alpha\}}
e^{-\beta\sum_\alpha (E_\alpha-\mu)n_\alpha}
=
\prod_\alpha \left(\sum_{n=0}^1 e^{-\beta(E_\alpha-\mu)n}\right)
=
\prod_\alpha \left(1+e^{-\beta(E_\alpha-\mu)}\right).
$$

Therefore,

$$
\langle n_\alpha\rangle
=
\frac{e^{-\beta(E_\alpha-\mu)}}{1+e^{-\beta(E_\alpha-\mu)}}
=
\frac{1}{e^{\beta(E_\alpha-\mu)}+1},
$$

which is the **Fermi--Dirac statistics**.



## Fermionic Creation and Annihilation Operators

Define fermionic ladder operators $\hat{c}_\alpha^\dagger$, $\hat{c}_\alpha$ and number operator

$$
\hat{n}_\alpha = \hat{c}_\alpha^\dagger \hat{c}_\alpha.
$$

For a single orbital, with basis states $\lvert 0\rangle$ and $\lvert 1\rangle$,

$$
\hat{c}^\dagger \lvert 0\rangle = \lvert 1\rangle,
\qquad
\hat{c}^\dagger \lvert 1\rangle = 0,
$$

$$
\hat{c}\lvert 1\rangle = \lvert 0\rangle,
\qquad
\hat{c}\lvert 0\rangle = 0,
$$

but since $n\in\{0,1\}$, one has

$$
(\hat{c}^\dagger)^2 = \hat{c}^2 = 0.
$$

In matrix form for one orbital,

$$
\hat{c} =
\begin{pmatrix}
0 & 1\\
0 & 0
\end{pmatrix},
\qquad
\hat{c}^\dagger =
\begin{pmatrix}
0 & 0\\
1 & 0
\end{pmatrix},
\qquad
\hat{n}=
\begin{pmatrix}
0 & 0\\
0 & 1
\end{pmatrix}.
$$

This implies

$$
\hat{c}\hat{c}^\dagger + \hat{c}^\dagger \hat{c} = \mathbb{I},
$$

i.e.

$$
\{\hat{c}^\dagger,\hat{c}\}=1.
$$

More generally, fermions satisfy **anticommutation relations** rather than commutation relations.



## Anticommutation for Different Orbitals

For different orbitals, one must be careful because exchanging fermions changes the sign of the state.

This leads to

$$
\hat{c}_x^\dagger \hat{c}_y^\dagger = -\, \hat{c}_y^\dagger \hat{c}_x^\dagger,
$$

and therefore

$$
\{\hat{c}_x^\dagger,\hat{c}_y^\dagger\}=0,
\qquad
\{\hat{c}_x,\hat{c}_y\}=0,
\qquad
\{\hat{c}_x^\dagger,\hat{c}_y\}=\delta_{xy}.
$$

Summary:

- **Bosons** $\rightarrow$ commutators
- **Fermions** $\rightarrow$ anticommutators



## Ordering Signs and the Jordan--Wigner String

Because an occupation state can be built as

$$
\lvert x_1,x_2,\dots\rangle_-
=
\hat{c}_{x_N}^\dagger \cdots \hat{c}_{x_2}^\dagger \hat{c}_{x_1}^\dagger \lvert 0\rangle
\qquad (x_1<x_2<\cdots),
$$

where $\lvert 0\rangle$ denotes the vacuum state with no particles. Changing the creation order introduces minus signs.

In particular,

$$
\hat{c}_\alpha^\dagger \lvert n_1,\dots,n_\alpha,\dots\rangle
=
\eta\,
\lvert n_1,\dots,n_\alpha+1,\dots\rangle,
$$

where the sign factor is

$$
\eta = (-1)^{\sum_{\beta<\alpha} n_\beta}.
$$

This sign factor is the same ordering structure that appears in the **Jordan--Wigner transformation**.

So even though the fermionic Hilbert space can look like a tensor product of spin-$\tfrac12$ degrees of freedom,
fermionic operators $\hat{c}_\alpha^\dagger$ and $\hat{c}_\alpha$ behave differently from ordinary spin raising and lowering operators. The full Jordan--Wigner transformation makes this relation explicit by mapping fermionic operators to nonlocal spin operators.



## Quick Comparison

| Feature | Bosons | Fermions |
|---|---|---|
| Exchange symmetry | symmetric | antisymmetric |
| Occupation numbers | $n_x = 0,1,2,\dots$ | $n_x = 0,1$ |
| Algebra | commutators | anticommutators |
| Thermal distribution | Bose--Einstein | Fermi--Dirac |
| Same-state occupancy | allowed | forbidden |
