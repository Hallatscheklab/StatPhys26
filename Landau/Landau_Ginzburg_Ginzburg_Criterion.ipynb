# Fluctuations and the Ginzburg Criterion

These notes extend Landau mean-field theory by allowing the order parameter to fluctuate in space. The main goal is to understand, from within the Landau framework itself, when mean-field theory becomes unreliable near a critical point.

## 1. Universality and the puzzle of mean-field theory

The liquid-gas critical point and the Ising critical point have the same type of singular behavior. For example, the density difference between liquid and gas behaves like the magnetization in an Ising magnet:

$$
\rho - \rho_c \sim (T_c - T)^\beta,
$$

and

$$
M \sim (T_c - T)^\beta.
$$

Experimentally, many systems share the same non-mean-field exponent, approximately

$$
\beta \approx 0.332.
$$

This is one of the first hints of **universality**: very different microscopic systems can have the same critical exponents. Landau theory, however, predicts mean-field exponents, such as

$$
\beta_{\rm MF} = \frac{1}{2}.
$$

So the question is: **Can we see, from inside a generalized Landau theory, why mean-field theory breaks down?**


## 2. From a single order parameter to an order-parameter field

In basic Landau theory, the order parameter is a single number, for example the total magnetization density

$$
\phi.
$$

The Landau free energy is expanded as

$$
\beta H_{\rm eff}(\phi)
= E_0 + h\phi + \frac{a}{2}\phi^2 + \frac{u}{4}\phi^4 + \cdots.
$$

This assumes that the thermodynamics is controlled by a spatially uniform order parameter.

Near a critical point, however, large domains fluctuate. We therefore introduce a slowly varying coarse-grained field

$$
\phi(\mathbf{x}).
$$

This field is not the microscopic spin at every lattice site. Rather, it is a smoothed version of the microscopic degrees of freedom over a length scale comparable to a coarse-graining scale.


## 3. Coarse graining

A concrete definition of a coarse-grained field is

$$
\phi(\mathbf{x})
= \sum_i \sigma_i\, w(\mathbf{x} - \mathbf{x}_i),
$$

where the smoothing kernel is normalized as

$$
\int d^D x\, w(\mathbf{x}) = 1.
$$

The kernel has width of order

$$
a,
$$

so by construction the field varies slowly on distances shorter than this scale.

Equivalently, the coarse-grained theory has a short-distance cutoff

$$
\Lambda \sim \frac{1}{a}.
$$

The exact effective Hamiltonian for the coarse-grained field is formally defined by summing over all microscopic configurations that produce a given field:

$$
\exp\!\left[-\beta H_{\rm eff}\{\phi(\mathbf{x})\}\right]
=
\frac{1}{Z_0}
\sum_{\{\sigma_i\}}
\exp[-\beta H(\{\sigma_i\})]
\prod_{\mathbf{x}}
\delta\!\left(\phi(\mathbf{x}) - \phi_{\rm cg}(\mathbf{x};\{\sigma_i\})\right).
$$

This is exact in principle, but not practical. Instead, we write the most general local expansion consistent with the symmetries.


## 4. Landau-Ginzburg free energy

For a slowly varying field, the effective Hamiltonian can be expanded in powers of the field and its gradients:

$$
\beta H_{\rm eff}[\phi]
=
\int d^D x\,
\left[
 h\phi
+ \frac{a}{2}\phi^2
+ \frac{u}{4}\phi^4
+ \frac{\kappa}{2}(\nabla \phi)^2
+ s_1\phi^2(\nabla\phi)^2
+ s_2(\nabla\phi)^4
+ \cdots
\right].
$$

Near the critical point, the order parameter is small and varies slowly. Keeping only the most important terms gives

$$
\beta H_{\rm eff}[\phi]
=
\int d^D x\,
\left[
\frac{a}{2}\phi^2(\mathbf{x})
+ \frac{\kappa}{2}(\nabla\phi)^2
+ \frac{u}{4}\phi^4(\mathbf{x})
+ \cdots
\right].
$$

The coefficient $a$ changes sign at the mean-field transition. For temperatures above the critical temperature, we take

$$
a(T) = \alpha t,
\qquad
 t = \frac{T - T_c}{T_c}.
$$


## 5. The Gaussian approximation

For now, ignore the nonlinear term $u\phi^4$. The effective Hamiltonian becomes

$$
\beta H_0[\phi]
=
\frac{1}{2}
\int d^D x\,
\left[
 a\phi^2(\mathbf{x}) + \kappa(\nabla\phi)^2
\right].
$$

This is the **Gaussian fixed point** or **Gaussian theory**.

The approximation is plausible when fluctuations are small, but we will have to check this assumption afterwards.


## 6. Fourier transform solution

Use the Fourier convention

$$
\phi(\mathbf{q})
= \frac{1}{\sqrt{V}}
\int d^D x\, e^{-i\mathbf{q}\cdot\mathbf{x}}\phi(\mathbf{x}).
$$

Then

$$
\nabla \rightarrow i\mathbf{q},
$$

and the Gaussian Hamiltonian becomes diagonal:

$$
\beta H_0
=
\frac{1}{2}
\sum_{\mathbf{q}}
\left(a + \kappa q^2\right)
\phi(\mathbf{q})\phi(-\mathbf{q}).
$$

Define

$$
\epsilon_q = a + \kappa q^2.
$$

Since $\phi(\mathbf{x})$ is real,

$$
\phi(\mathbf{q}) = \phi^*(-\mathbf{q}).
$$

Therefore modes $\mathbf{q}$ and $-\mathbf{q}$ are not independent; they should be grouped together in the Gaussian integral.


For a complex Gaussian variable $z=x+iy$,

$$
\int dx\,dy\,e^{-\epsilon |z|^2}=\frac{\pi}{\epsilon}.
$$

The Gaussian free energy is therefore, up to unimportant constants,

$$
F
= -\frac{1}{\beta}\ln Z
= -\frac{k_B T}{2}\sum_{\mathbf{q}}
\ln\!\left(\frac{\pi k_B T}{\epsilon_q}\right).
$$

Equivalently, the singular part is controlled by

$$
F_{\rm sing}
\sim
\frac{k_B T}{2}\sum_{\mathbf{q}}
\ln\left(a+\kappa q^2\right).
$$


## 7. Correlation function

The two-point correlation function is

$$
C(\mathbf{r})
= \langle \phi(\mathbf{r})\phi(\mathbf{0})\rangle.
$$

Using the Fourier expansion,

$$
C(\mathbf{r})
= \frac{1}{V}
\sum_{\mathbf{q},\mathbf{q}'}
e^{i\mathbf{q}\cdot\mathbf{r}}
\left\langle \phi(\mathbf{q})\phi(\mathbf{q}')\right\rangle.
$$

By translation invariance, the only nonzero contribution comes from

$$
\mathbf{q}'=-\mathbf{q}.
$$

For the Gaussian theory,

$$
\left\langle \phi(\mathbf{q})\phi(-\mathbf{q})\right\rangle
=
\frac{1}{\beta\epsilon_q}
=
\frac{1}{\beta(a+\kappa q^2)}.
$$

Thus

$$
C(\mathbf{r})
= \frac{1}{\beta V}
\sum_{\mathbf{q}}
\frac{e^{i\mathbf{q}\cdot\mathbf{r}}}{a+\kappa q^2}.
$$

In the thermodynamic limit,

$$
C(\mathbf{r})
=
\frac{1}{\beta}
\int\frac{d^D q}{(2\pi)^D}
\frac{e^{i\mathbf{q}\cdot\mathbf{r}}}{a+\kappa q^2}.
$$


## 8. Correlation length

Rewrite the denominator as

$$
a + \kappa q^2
= \kappa\left(q^2 + \xi^{-2}\right),
$$

where

$$
\xi = \sqrt{\frac{\kappa}{a}}.
$$

The length $\xi$ is the **correlation length**.

At criticality, $a\to 0$ and therefore

$$
\xi \to \infty.
$$

At $a=0$, the correlation function has a power-law form:

$$
\beta C(r) \propto \frac{1}{\kappa r^{D-2}}.
$$

Away from criticality, correlations decay exponentially. In three dimensions,

$$
\beta C(r)
=
\frac{1}{4\pi\kappa}\frac{e^{-r/\xi}}{r}.
$$

More generally, at large $r$ and fixed $\xi$,

$$
C(r) \propto \frac{e^{-r/\xi}}{r^{(D-1)/2}}.
$$


```python
import numpy as np
import matplotlib.pyplot as plt

r = np.linspace(0.05, 10, 400)
xis = [0.5, 1.0, 2.0, 5.0]

plt.figure(figsize=(6, 4))
for xi in xis:
    C = np.exp(-r/xi)/r
    plt.plot(r, C, label=fr"$\xi={xi}$")
plt.yscale("log")
plt.xlabel(r"distance $r$")
plt.ylabel(r"$e^{-r/\xi}/r$ (3D form, arbitrary units)")
plt.title("Exponential decay of correlations away from criticality")
plt.legend()
plt.show()
```

Since

$$
a(T)=\alpha t,
$$

we obtain

$$
\xi = \sqrt{\frac{\kappa}{\alpha}}\, t^{-1/2}.
$$

Therefore the Gaussian theory predicts the mean-field correlation-length exponent

$$
\xi \sim t^{-\nu},
\qquad
\nu_{\rm MF}=\frac{1}{2}.
$$


## 9. Fluctuation contribution to the heat capacity

The singular part of the Gaussian free energy gives a fluctuation contribution to the heat capacity.

The heat capacity is

$$
C = -T\frac{\partial^2 F}{\partial T^2}.
$$

Since

$$
\epsilon_q = a(T) + \kappa q^2,
$$

and

$$
a(T)=\alpha t,
$$

one finds, up to nonsingular prefactors,

$$
\frac{C}{V}
\sim
\frac{k_B}{2}
\int\frac{d^Dq}{(2\pi)^D}
\left(\frac{\alpha}{a+\kappa q^2}\right)^2.
$$

Equivalently,

$$
\frac{C}{V}
\sim
\frac{k_B\alpha^2}{2\kappa^2}
\int\frac{d^Dq}{(2\pi)^D}
\frac{1}{(q^2+\\xi^{-2})^2}.
$$


## 10. UV versus IR divergences

The integral can fail to converge for two distinct reasons.

### Ultraviolet divergence

Large $q$ corresponds to short distances. The continuum Gaussian theory incorrectly includes fluctuations down to arbitrarily small length scales.

But the coarse-grained field was only meant to vary slowly on scales larger than $a$. Therefore we introduce a cutoff

$$
|\mathbf{q}| < \Lambda,
\qquad
\Lambda \sim \frac{1}{a}.
$$

Dependence on $\Lambda$ is not universal, because it depends on microscopic details.

### Infrared divergence

Small $q$ corresponds to long distances. As $t\to 0$, the correlation length diverges and long-wavelength modes become increasingly important. This is the physically important critical divergence.


## 11. Scaling of the fluctuation heat capacity

With the cutoff included,

$$
\frac{C}{V}
\sim
\frac{k_B\alpha^2}{2\kappa^2}
\int_0^\Lambda dq\,
\frac{q^{D-1}}{(q^2+\xi^{-2})^2}.
$$

Rescale momenta by

$$
\bar q = q\xi.
$$

Then

$$
\frac{C}{V}
\sim
\frac{k_B\alpha^2}{2\kappa^2}
\xi^{4-D}
\int_0^{\xi\Lambda} d\bar q\,
\frac{\bar q^{D-1}}{(1+\bar q^2)^2}.
$$

For large $\xi\Lambda$, the integral behaves as

$$
\int_0^{\xi\Lambda} d\bar q\,
\frac{\bar q^{D-1}}{(1+\bar q^2)^2}
\sim
\begin{cases}
(\xi\Lambda)^{D-4}, & D>4,\\[4pt]
\ln(\xi\Lambda), & D=4,\\[4pt]
\text{constant}, & D<4.
\end{cases}
$$

Therefore the singular fluctuation contribution behaves as

$$
\frac{C}{V}
\sim
\begin{cases}
\text{nonsingular cutoff-dependent constant}, & D>4,\\[4pt]
\ln(\xi/a), & D=4,\\[4pt]
\xi^{4-D}, & D<4.
\end{cases}
$$


Because

$$
\xi \sim t^{-1/2},
$$

for $D<4$ the fluctuation contribution scales as

$$
C_{\rm fluc} \sim t^{-(4-D)/2}.
$$

Thus, for $D\le 4$, fluctuations produce a singular contribution to thermodynamics. This identifies

$$
D_c = 4
$$

as the **upper critical dimension** of the scalar $\phi^4$ theory.

Above four dimensions, mean-field theory is self-consistent near the critical point. At and below four dimensions, fluctuations increasingly modify mean-field predictions.


## 12. The Ginzburg criterion

Mean-field theory is valid when fluctuation corrections are small compared with the mean-field result. A convenient way to estimate this is to ask when the fluctuation contribution becomes order one.

Let

$$
R_* = \sqrt{\frac{\kappa}{\alpha}} = \xi(t=1).
$$

The Ginzburg temperature scale is obtained from the condition that the fluctuation correction becomes comparable to the mean-field contribution. The result is

$$
t_G
\sim
\left(\frac{a}{R_*}\right)^{\frac{2D}{4-D}}.
$$

Here $a$ is the microscopic coarse-graining scale.

For

$$
t > t_G,
$$

fluctuations are relatively small and mean-field theory is approximately valid.

For

$$
t < t_G,
$$

fluctuations dominate and mean-field theory must be corrected.


## 13. Physical picture

For $T>T_c$, the system is organized into fluctuating domains of size approximately

$$
\xi.
$$

As the critical point is approached,

$$
\xi \to \infty.
$$

The system no longer consists of small, weakly correlated patches. Instead, fluctuations occur on all length scales. This is why a local Taylor expansion in a single uniform order parameter is not enough.

At criticality, there is no characteristic finite domain size. The correlations become scale-free:

$$
C(r) \sim \frac{1}{r^{D-2}}
$$

in the Gaussian theory.


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
N = 120
xis = [1.5, 4.0, 10.0]

kx = 2 * np.pi * np.fft.fftfreq(N)
ky = 2 * np.pi * np.fft.rfftfreq(N)
KX, KY = np.meshgrid(kx, ky, indexing='ij')
q2 = KX**2 + KY**2

def gaussian_field(xi):
    # Use the Gaussian Landau-Ginzburg structure factor S(q) ~ 1/(xi^-2 + q^2).
    spectrum = 1.0 / (xi**-2 + q2)
    spectrum[0, 0] = 0.0
    noise = rng.normal(size=(N, N // 2 + 1)) + 1j * rng.normal(size=(N, N // 2 + 1))
    field = np.fft.irfft2(np.sqrt(spectrum) * noise, s=(N, N))
    field -= field.mean()
    field /= field.std()
    return field

fields = [gaussian_field(xi) for xi in xis]
vmax = max(np.max(np.abs(f)) for f in fields)

fig, axes = plt.subplots(1, 3, figsize=(11, 3.6), constrained_layout=True)
for ax, xi, field in zip(axes, xis, fields):
    im = ax.imshow(field, cmap='coolwarm', vmin=-vmax, vmax=vmax, origin='lower')
    ax.set_title(fr'$\xi={xi}$')
    ax.set_xticks([])
    ax.set_yticks([])

fig.colorbar(im, ax=axes, shrink=0.85, label=r'field $\phi(\mathbf{x})$')
fig.suptitle(r'Gaussian random-field snapshots: larger $\xi$ means larger correlated domains', y=1.04)
plt.show()
```

These snapshots are sampled from a Gaussian random field with structure factor
$$
S(q) \propto \frac{1}{\xi^{-2}+q^2}.
$$
They are not a full interacting Ising simulation, but they capture the key physical message of the Gaussian Landau-Ginzburg theory: as $\xi$ grows, the field organizes into larger correlated patches, and near criticality there is no short finite length scale left.


## 14. Summary

The Landau-Ginzburg theory replaces the uniform order parameter by a fluctuating field:

$$
\phi \longrightarrow \phi(\mathbf{x}).
$$

The Gaussian approximation is exactly solvable and gives

$$
\left\langle \phi(\mathbf{q})\phi(-\mathbf{q})\right\rangle
=
\frac{1}{\beta(a+\kappa q^2)}.
$$

The correlation length is

$$
\xi = \sqrt{\frac{\kappa}{a}}
\sim t^{-1/2}.
$$

The fluctuation correction to the heat capacity scales as

$$
C_{\rm fluc}
\sim
\begin{cases}
\text{finite}, & D>4,\\[4pt]
\ln(1/t), & D=4,\\[4pt]
t^{-(4-D)/2}, & D<4.
\end{cases}
$$

Thus $D_c=4$ is the upper critical dimension for the scalar Landau-Ginzburg theory. For dimensions below four, sufficiently close to the critical point, fluctuations invalidate the naive mean-field approximation.
